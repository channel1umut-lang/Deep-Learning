```python
# Stacking
```

```python
x
```

```python
x.shape
```

```python
torch.stack([x,x,x,x], dim=0) # 4 tane x' i aldi ust uste koydu
#4 → en dıştaki grup sayısı
#2 → her grubun içindeki alt grup sayısı
#3 → en içteki eleman sayısı
```

```python
torch.stack([x,x,x,x], dim=1) # dim=1 → yeni boyutun yeri, 4 → liste uzunluğu. Yani Yeni boyutu index 1’e koy demek. Yani torch.Size([2, ? ,3]) soru isareti yerine 4 gelir. Burda 2 = 0 inci index
#“2 tane şey var her birinin içinde 4 tane şey var her birinin içinde 3 eleman var”
#NOTE!!!!!!!!!
#2 → en dıştaki grup sayısı
#4 → her grubun içindeki alt grup sayısı
#3 → en içteki eleman sayısı
```

```python
torch.stack([x,x,x,x], dim=2) # torch.Size([2, 3, ?]). Burda soru isaretinin yeri yeni boyutumun indexinin yerini gosteriyor gosteriyor.
#2 → en dıştaki grup sayısı
#3 → her grubun içindeki alt grup sayısı
#4 → en içteki eleman sayısı
```
